# handing missing values

### Handling Missing Values in Pandas

Pandas represents missing values using `NaN`.

| Method | Purpose |
|---|---|
| `s.isnull()` | Checks missing values |
| `s.notnull()` | Checks non-missing values |
| `s.isnull().sum()` | Counts missing values |
| `s.dropna()` | Removes missing values |
| `s.fillna(0)` | Replaces missing values with 0 |

```python
f = [12,None ,13,None,15]
data = pd.Series(f,index=['a','b','c','d','e'])
data
```

```python
data.isnull()
```

```python
data.notnull()
```

```python
data.isnull().sum()
```

```python
data.dropna() #remove missing values 
```

```python
data.fillna(0) # replace missing value with no insidethe brackets.
```

```python
data.fillna(9)
```

# Important: These methods generally return a new Series; they don't permanently change the original s unless you assign the result back.

| Method | Purpose |

| `sort_index()` | Sorts indexes |
