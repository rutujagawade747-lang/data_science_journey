1. What is a DataFrame?
### A DataFrame is a two-dimensional data structure in Pandas that stores data in rows and columns.


```python
import pandas as pd 
```

```python
data ={
    "Name":["riya","khushi","neha"],
    "age":[20,21,28],
    "marks":[85,90,78]
}
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df
```

# Understand this code
- pd.DataFrame(data) creates a DataFrame.
- Dictionary keys become column names.
- Dictionary values become column data.
- Pandas automatically creates indexes starting from 0.

```python
data ={
    "Name":["aditi","sneha","kabya","anaya"],
    "age":[20,21,19,20],
    "marks":[88,92,85,90]
}
```

```python
df = pd.DataFrame(data)
df
```

## Pandas DataFrame Attributes

DataFrame attributes provide information about the structure and
properties of a DataFrame.

| Attribute | Description |
|---|---|
| df.shape | Returns the number of rows and columns. |
| df.size | Returns the total number of elements. |
| df.ndim | Returns the number of dimensions. |
| df.columns | Returns all column names. |
| df.index | Returns row index labels. |
| df.dtypes | Returns the data type of each column. |
| df.values | Returns data as a NumPy array. |

Important Points:
1. A DataFrame is a 2-dimensional data structure.
2. df.shape returns (rows, columns).
3. df.size = number of rows × number of columns.
4. df.ndim returns 2 for a DataFrame.
5. Attributes do not require parentheses.
