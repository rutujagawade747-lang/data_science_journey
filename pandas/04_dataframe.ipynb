## Pandas DataFrame Attributes

properties of a DataFrame.

| Attribute | Description |
|---|---|
| df.shape | Returns the number of rows and columns. |
| df.size | Returns the total number of elements. |
| df.ndim | Returns the number of dimensions. |
| df.columns | Returns all column names. |
| df.index | Returns row index labels. |
| df.dtypes | Returns the data type of each column. |
| df.values | Returns data as a NumPy array. |

Important Points:
1. A DataFrame is a 2-dimensional data structure.
2. df.shape returns (rows, columns).
3. df.size = number of rows × number of columns.

## Accessing Rows in Pandas DataFrame

### 1. loc[]
- Used for label-based selection.
- Syntax: df.loc[row_label]
- Selects rows using index labels.
- Ending label is included in slicing.
- Multiple rows: df.loc[[0, 2]]
- Rows and columns: df.loc[0:2, ["Name", "Marks"]]

### 2. iloc[]
- Used for position-based selection.
- Syntax: df.iloc[position]
- Selects rows using integer positions.
- Ending position is excluded in slicing.
- Multiple rows: df.iloc[[0, 2]]
- Rows and columns: df.iloc[0:2, 0:2]

### Difference between loc[] and iloc[]

| loc[] | iloc[] |
|---|---|
| Label-based | Position-based |
| Uses index labels | Uses integer positions |
| Ending label included | Ending position excluded |

```python
df.loc[1]
```

```python
df.iloc[2]
```

```python
df.loc[[0,3]] # 0 and 3 so two brackets
```

```python
df.iloc[0:3]
```

```python
df.loc[0:2,["Name","Age"]]
```

```python
df.iloc[0:2,0:2]
```

# Remember this trick:
- loc = Labels (includes the end).
- iloc = Integer positions (excludes the end).

## Conditional Filtering in Pandas

Conditional filtering is used to select rows based on specified conditions.

Syntax:
df[df["column_name"] > value]

Comparison Operators:
-  | > : Greater than
-  <  : Less than
-  | >= : Greater than or equal to
- <= : Less than or equal to
- == : Equal to
- != : Not equal to

Examples:
- df[df["Marks"] > 85]
- df[df["Age"] == 20]
- df[df["Marks"] < 90]

## Filtering Using Multiple Conditions

1. AND (&):
   Both conditions must be true.
   > df[(df["Marks"] > 85) & (df["Age"] == 20)]

2. OR (|):
   At least one condition must be true.
   > df[(df["Marks"] > 90) | (df["Age"] == 19)]

3. NOT (~):
   Reverses the condition.
   > df[~(df["Marks"] > 85)]

Important:
- Always put each condition inside parentheses.
- Use & instead of and.
- Use | instead of or.
- Use ~ instead of not.

```python
df[(df["Marks"]>85) &(df["Age"]==20) ]
```

```python
df[(df["Marks"]>90) | (df["Age"]==19) ]
```

```python
df[(df["Marks"]<92) & (df["Age"]>19) ]
```

```python
df[~(df["Marks"]>88) ]
```

## Additional Filtering – isin()

- isin() checks whether values belong to a given list.
- It returns True for matching values.

Syntax:
df[df["column"].isin([value1, value2])]

Example:
df[df["Age"].isin([19, 21])]

```python
df[df["Age"].isin([19, 21])]
```

```python
df[df["Age"].isin([19,20])]
```

Syntax:
