```python
import numpy as np

students = np.array([
    [101, 78, 85, 72],
    [102, 45, 67, 55],
    [103, 92, 88, 95],
    [104, 38, 49, 44],
    [105, 81, 76, 89],
    [106, 60, 52, 65],
    [107, 95, 91, 90],
    [108, 73, 69, 80],
    [109, 48, 58, 51],
    [110, 87, 79, 84]
])
```

```python
print(np.sum(students[:,1:],axis=1))
```

```python
print(np.mean(students[:,1:],axis=1))
```

```python
avg = np.mean(students[:,1:],axis=1)
print(np.where(avg>=50,"passs","fail"))
```

```python
avg = np.mean(students[:,1:],axis=1)
max1 =np.argmax(avg)
print(students[max1,0])
```

# np.argsort(avg)       # lowest → highest
 # np.argsort(avg)[::-1] # highest → lowest

```python
ind =np.argsort(avg)[::-1]
print(students[ind])
```

```python
print(np.mean(students[:,1:],axis=0))
```

```python
print(np.max(students[:,1:],axis=0))
```

```python
print(np.min(students[:,1:],axis=0))
```

```python
ind = students[:,1:] >= 50
passed = np.all(ind, axis=1)
print(passed)
print(np.sum(passed))
```

✅ np.any() → at least one subject is below 50.


np.all() → everything must be True
np.any() → at least one must be True

```python
ind = students[:,1:]<50
failed = np.any(ind,axis=1)
print(failed)
print(np.sum(failed))
```

for reverse [::-1]

[sorted[:3]] this used when one is in text and we have no 


```python
avg_1 = np.mean(students[:,1:],axis=1)
sorted = np.argsort(avg_1)[::-1]
print(students[sorted[:3]])
```

```python
ind =  students[:,1]
student_id = np.argmax(ind,axis =0)
print(students[student_id,0])
```

```python
allsub= np.sum(students[:,1:],axis=0)
print(np.sum(allsub))
```

```python
highest = np.sum(students[:,1:],axis=0)
hiest_sub=np.argmax(highest)
print(hiest_sub)
```
