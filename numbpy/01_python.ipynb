```python
text = "hello"#range(start, stop, step)
for i in range(len(text)-1,-1,-1):
    print(text[i])

 
```

## Q3 even numbers

```python
numbers = [10, 25, 7, 42, 18, 31, 8]
count=0
for i in numbers:
    if i%2==0:
        count=count+1

print(count)
```

## Q4 claculate frequency

```python
numbers = [1, 2, 2, 3, 1, 4, 2, 3]

frequency = {}
for i in numbers:
    if i in frequency:
        frequency[i]=frequency[i]+1
    else:      
      frequency[i]=1   


for j in frequency: 
    print(frequency[j]) 
```
