# NumPy Easy Rule 🧠

## 1. `np.function()`
Use `np.` when NumPy provides the function.

Examples:
- `np.array()`
- `np.where()`
- `np.delete()`
- `np.sort()`
- `np.concatenate()`

👉 **NumPy does the work → `np.`**

---

## 2. `array.method()`
Use the array variable when the array itself has the method.

Examples:
- `arr.sum()`
- `arr.mean()`
- `arr.max()`
- `arr.reshape()`
- `arr.sort()`

👉 **Array does the work → `arr.`**

---

## 3. Important ⚠️

`np.` does NOT always mean "new array."

Check the specific function.

Example:

`np.sort(arr)` → returns sorted copy  
`arr.sort()` → changes original array

---

## 🧠 One-line memory trick

**"NumPy tool → `np.` | Array tool → `arr.`"**

# delete operation

```python
arr=np.array([1,2,3,4,5])
delete = np.delete(arr,4)
print(delete)
```

## - axis=None — flatten the array and delete an element by its flattened index.
# - axis=0 — delete a row.
# - axis=1 — delete a column.
# Also, np.delete() returns a new array; it doesn't modify arr2 itself.


```python
arr2=np.array([[1,2],
               [3,5],
               [7,8]])
delete=np.delete(arr2,[1][0]) # ithe tyni arryla flatt kela mag index 1 valayla delete kela
print(delete)
```

```python
arr2=np.array([[1,2],
               [3,5],
               [7,8]])
delete=np.delete(arr2,1,axis=0) # ithe tyni axis wise gela
print(delete)
```

```python
arr2=np.array([[1,2],
               [3,5],
               [7,8]])
delete=np.delete(arr2,0,axis=1)
print(delete)
```
