```python
import numpy as np

python_marks = np.array([
    [101, 78],
    [102, 92],
    [103, 65],
    [104, 88]
])

sql_marks = np.array([
    [101, 85],
    [102, 76],
    [103, 90],
    [104, 81]
])
```

Why the double parentheses?
- Outer () → calls the np.hstack() function.
- Inner () → groups python_marks and sql_marks into one tuple.

# np.hstack((array1, array2))  # Horizontal
 # np.vstack((array1, array2))  # Vertical
 # np.concatenate((array1, array2), axis=0)  # Vertical by default

```python

last=np.hstack((python_marks,sql_marks))
print(last)
```

```python
new_student = np.array([[105, 95]])
python_marks=np.vstack((python_marks,new_student))
print(python_marks)
```

# np.unique()
### 🔹 What does np.unique() do?
It returns the unique (non-repeated) values in an array.
It also sorts the unique values by default.

```python
mark=np.array([[10,3,4],
               [5,3,5],
               [8,7,9]])
```

```python
print(np.unique(mark))
```

```python
courses = np.array([
    "Python",
    "SQL",
    "Python",
    "NumPy",
    "SQL",
    "Pandas",
    "Python"
])

print(np.unique(courses))
```

```python
values, counts = np.unique(courses, return_counts=True)

print(values)
print(counts)
#values = What are the different things?
#counts = How many times does each thing appear?
```

# np.random.seed()
### A seed lets us make the random results repeatable.

```python
np.random.seed(42)
print(np.random.randint(1,50,5))
```

So remember:
Seed = controls/repeats the random sequence, NOT the minimum random number.

```python
np.random.seed(50)

numbers = np.random.randint(1, 100, 5)
print(numbers)
```

# copy vs view

### view 

```python
a = np.array([10,20,30,40])
b= a.view()
b[0]=100
print(a)
print(b)
```

# view can change original array 
# but copy didtn change the original array it create copy 

```python
a = np.array([10, 20, 30, 40])

b = a.copy()

b[0] = 100

print(a)
print(b)
```
