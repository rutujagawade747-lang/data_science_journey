```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
```

```python
print("=== zomatos sales analysis ==")
print("\n sales data shape",sales_data.shape)
# print("\n sample data for 1st 3 restraunt \n",sales_data[0:3])
print("\n sample data for 1st 3 restraunt \n",sales_data[:3]) 


```

```python
print("\n sample data for 1st 3 restraunt \n",sales_data[:,1:])# sagle rows yenr pn colum
# one pasun suruhone
```

```python
# total sales per year 
print(np.sum(sales_data[:,1:],axis=0))
# total sales per restraunta
print(np.sum(sales_data[:,1:],axis=1))
```

```python
# maximum sales per year 
print(np.max(sales_data[:,1:],axis=0))
```

```python
# average sales per resrauntant 
print(np.mean(sales_data[:,1:],axis=1))
```

# Easy way to remember
Cumulative = running total 🔄
cumsum = cumulative sum.

```python
cusum=np.cumsum(sales_data[:,1:],axis=1)
print(cusum)
```

```python
plt.figure(figsize=(10,4))
plt.plot(np.mean(cusum,axis=0))
plt.title("average cumulative sales acros all restraunt ")
plt.xlabel("years")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```
